# Assignment 7.1 (Taxi3)

```python
import gymnasium as gym
import pygame
import time

# Initialize pygame properly
pygame.init()

# Create Taxi environment with human rendering (GUI window)
env = gym.make("Taxi-v3", render_mode="human")
obs, _ = env.reset()

print("🚖 Taxi-v3 Manual Control Started!")
print("Controls:")
print("  ↑ Up Arrow    = Move North")
print("  ↓ Down Arrow  = Move South")
print("  ← Left Arrow  = Move West")
print("  → Right Arrow = Move East")
print("  P             = Pickup passenger")
print("  D             = Dropoff passenger")
print("  ESC or Close Window = Quit")
print("  R             = Reset episode")
print("=" * 40)

# Map keys to Taxi actions
KEY_TO_ACTION = {
    pygame.K_UP: 1,     # North
    pygame.K_DOWN: 0,   # South
    pygame.K_RIGHT: 2,  # East
    pygame.K_LEFT: 3,   # West
    pygame.K_p: 4,      # Pickup
    pygame.K_d: 5,      # Dropoff
}

clock = pygame.time.Clock()
running = True
action = None  # No action initially
step_count = 0
episode_count = 1
total_reward = 0

try:
    while running:
        # Process events (keyboard and window close)
        for event in pygame.event.get():
            if event.type == pygame.QUIT:
                running = False
            elif event.type == pygame.KEYDOWN:
                if event.key == pygame.K_ESCAPE:
                    running = False
                elif event.key == pygame.K_r:
                    print("🔄 Manual reset requested")
                    obs, _ = env.reset()
                    step_count = 0
                    total_reward = 0
                    episode_count += 1
                    print(f"🆕 Episode {episode_count} started")
                elif event.key in KEY_TO_ACTION:
                    action = KEY_TO_ACTION[event.key]

        # If no action key pressed, do nothing (action=None means no step)
        if action is not None:
            obs, reward, terminated, truncated, info = env.step(action)
            step_count += 1
            total_reward += reward

            print(f"Step {step_count:3d} | Action: {action} | Reward: {reward} | Total: {total_reward}")

            # Render is automatic in render_mode="human"

            if terminated or truncated:
                print("\n" + "=" * 40)
                print(f"Episode {episode_count} ended.")
                print(f"Total steps: {step_count}")
                print(f"Total reward: {total_reward}")
                print("=" * 40)
                print("🔄 Resetting in 3 seconds... (Press R to reset immediately)")

                start_time = time.time()
                reset_now = False
                while time.time() - start_time < 3 and not reset_now:
                    for event in pygame.event.get():
                        if event.type == pygame.QUIT:
                            running = False
                            reset_now = True
                        elif event.type == pygame.KEYDOWN:
                            if event.key == pygame.K_ESCAPE:
                                running = False
                                reset_now = True
                            elif event.key == pygame.K_r:
                                reset_now = True
                    clock.tick(60)

                if running:
                    obs, _ = env.reset()
                    step_count = 0
                    total_reward = 0
                    episode_count += 1
                    print(f"🆕 Episode {episode_count} started")
            
            # Reset action after each step
            action = None

        # Control frame rate
        clock.tick(30)  # 30 FPS

except KeyboardInterrupt:
    print("\n🛑 Interrupted by user (Ctrl+C)")
except Exception as e:
    print(f"❌ Runtime error: {e}")
finally:
    print("🔚 Closing environment...")
    env.close()
    pygame.quit()
    print("✅ Cleanup complete!")
```

```python
import gymnasium as gym

# Create Taxi-v3 environment with text rendering
env = gym.make("Taxi-v3", render_mode="ansi")

print("🚖 Taxi-v3 Environment")
print("Description: Pick up and drop off passengers in a grid world.")
print(f"Number of states: {env.observation_space.n}")  # 500
print(f"Number of actions: {env.action_space.n}")      # 6

print("\nReward structure:")
print("  -1 for each step")
print("  +20 for successful dropoff")
print("  -10 for illegal pickup/dropoff")

print("\nDeterministic environment: Yes, outcomes fixed by action")
print("="*60)

# Reset environment
state, _ = env.reset()
print(f"Initial state: {state}")

# Get initial rendered environment string
render_str = env.render()
print(render_str)

# Extract passenger and destination from the last line of the render output (optional)
last_line = render_str.strip().split('\n')[-1]
if "Passenger at" in last_line:
    print(last_line)

print("="*60)

# Actions with their names for clarity
action_names = ['South', 'North', 'East', 'West', 'Pickup', 'Dropoff']

# Example manual action sequence:
# Pickup passenger at R, move East twice, then Dropoff at G
actions = [4, 2, 2, 5]

total_reward = 0
done = False

print("Starting manual/hardcoded action loop...\n")

for step, action in enumerate(actions, 1):
    state, reward, terminated, truncated, info = env.step(action)
    total_reward += reward

    print(f"Step {step}: Action = {action} ({action_names[action]})")
    render_str = env.render()
    print(render_str)
    
    # Check if passenger was picked up or dropped off based on action and reward
    if action == 4 and reward == -1:
        print("Passenger picked up")
    elif action == 5 and reward > 0:
        print("Passenger dropped off successfully")
    elif action == 5 and reward < 0:
        print("Illegal dropoff attempt")
        
    print(f"Reward received: {reward}")
    print("-" * 40)

    if terminated or truncated:
        print("🚩 Episode ended")
        break

print(f"\n🏁 Total reward after {step} steps: {total_reward}")
print("="*60)

# Close environment
env.close()
```
