```python
import gymnasium as gym
import time

# Create environment with GUI rendering
env = gym.make("Taxi-v3", render_mode="human")

# Reset environment
state, _ = env.reset()
print("Initial State:", state)

# Step through manually chosen actions
actions = [1, 2, 0, 4, 5]  # Right, Down, South, Pickup, Drop-off
for action in actions:
    next_state, reward, terminated, truncated, info = env.step(action)
    print(f"Action: {action}, Next State: {next_state}, Reward: {reward}")
    time.sleep(1)  # pause to see step-by-step visually
    if terminated or truncated:
        break

env.close()
```
