```python
import gymnasium as gym

# Create Taxi-v3 environment with text rendering
env = gym.make("Taxi-v3", render_mode="ansi")

# Print environment description
print("🚖 Taxi-v3 Environment")
print("Description: Pick up and drop off passengers in a grid world.")
print("Number of states:", env.observation_space.n)  # 500
print("Number of actions:", env.action_space.n)      # 6

print("\nReward structure:")
print("-1 for each step, +20 for successful dropoff, -10 for illegal pickup/dropoff")

print("\nDeterministic environment: Yes, outcomes fixed by action")

# Reset environment
state, _ = env.reset()
print("\nInitial state:", state)

# Render initial state as text
print(env.render())

# Hardcoded simple manual action sequence (pickup and dropoff)
actions = [5, 2, 2, 4, 4, 3, 1]  # example actions (could be pickup, dropoff, move)

total_reward = 0
done = False

print("\nStarting manual/hardcoded action loop...")

for step, action in enumerate(actions):
    state, reward, terminated, truncated, info = env.step(action)
    total_reward += reward
    print(f"\nStep {step+1}: Action = {action}")
    print(env.render())
    print(f"Reward received: {reward}")
    if terminated or truncated:
        print("Episode ended")
        break

print(f"\nTotal reward after {step+1} steps: {total_reward}")

# Close environment
env.close()
```
